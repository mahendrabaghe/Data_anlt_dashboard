```python
import sys
sys.path.insert(0, '../src')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from analytics import load_data, calculate_kpis, monthly_sales, category_performance, generate_insights
from customer_segmentation import run_segmentation
sns.set_theme(style='darkgrid', palette='husl')
df = load_data()
print(f'Records loaded: {len(df):,}')
```

## 1. Dataset Overview

```python
df.head(10)
```

```python
df.info()
df.describe()
```

```python
print('Missing values:')
print(df.isnull().sum())
```

## 2. Key Performance Indicators

```python
kpis = calculate_kpis(df)
for k, v in kpis.items():
    print(f'{k:<30} {v}')
```

## 3. Monthly Revenue and Profit Trend

```python
mon = monthly_sales(df)
fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(mon['Month'], mon['Revenue'], marker='o', label='Revenue', linewidth=2)
ax.plot(mon['Month'], mon['Profit'], marker='s', label='Profit', linewidth=2, linestyle='--')
ax.set_title('Monthly Revenue and Profit', fontsize=14, fontweight='bold')
ax.tick_params(axis='x', rotation=45)
ax.legend()
plt.tight_layout()
plt.show()
```

## 4. Category Performance

```python
cat = category_performance(df)
fig = px.bar(cat, x='Category', y='Revenue', color='Profit',
             color_continuous_scale='Plasma', title='Revenue by Category')
fig.show()
```

## 5. Sales by State

```python
from analytics import state_performance
state = state_performance(df)
fig = px.bar(state, x='Revenue', y='State', orientation='h', color='Revenue',
             color_continuous_scale='Teal', title='Revenue by State')
fig.show()
```

## 6. Gender and Age Group Analysis

```python
from analytics import gender_analysis, age_group_analysis
gen = gender_analysis(df)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].pie(gen['Revenue'], labels=gen['Gender'], autopct='%1.1f%%')
axes[0].set_title('Revenue by Gender')
ag = age_group_analysis(df)
axes[1].bar(ag['Age_Group'].astype(str), ag['Revenue'], color='steelblue')
axes[1].set_title('Revenue by Age Group')
plt.tight_layout()
plt.show()
```

## 7. Discount vs Profit Margin

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df['Discount']*100, df['Profit_Margin'], alpha=0.3, s=8, c='steelblue')
ax.axhline(0, color='red', linestyle='--')
ax.set_title('Discount vs Profit Margin')
ax.set_xlabel('Discount (%)')
ax.set_ylabel('Profit Margin (%)')
plt.show()
```

## 8. Customer Segmentation (RFM)

```python
rfm_df, summary = run_segmentation(df)
print(summary.to_string(index=False))
```

```python
fig = px.scatter(rfm_df, x='Recency', y='Monetary', size='Frequency',
                 color='Segment', opacity=0.7, title='RFM Scatter Plot')
fig.show()
```

## 9. Correlation Heatmap

```python
cols = ['Sales', 'Cost', 'Profit', 'Quantity', 'Unit_Price', 'Discount', 'Profit_Margin']
fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(df[cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', ax=ax)
ax.set_title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
